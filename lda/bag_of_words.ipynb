```python
import json
import string

# Read the JSON file
with open('kenny_transcripts_modified.json', 'r') as file:
    data = json.load(file)

# Extract the transcripts, remove newline characters, and punctuation except for apostrophe and percentage sign
transcripts = [transcript.replace('\n', '') for transcript in data['Transcripts']]
transcripts = [''.join(char for char in transcript if char not in string.punctuation or char in ["'", '%']) for transcript in transcripts]

# Tokenize each transcript into words, make them lowercase, and join with commas
bag_of_words = [','.join(word.lower() for word in transcript.split()) for transcript in transcripts]

# Save the bag of words into a new JSON file
output_data = {'BagOfWords': bag_of_words}
with open('kenny_bag_of_words.json', 'w') as output_file:
    json.dump(output_data, output_file)
```

```python
import json

# Load the JSON data
with open('kenny_transcripts.json', 'r') as file:
    data = json.load(file)

# Replace '19' with 'nineteen' and '26' with 'twenty-six' in the transcript
transcript = data['Transcripts'][0]
transcript = transcript.replace('19', 'nineteen').replace('26', 'twenty-six')

# Update the transcript in the data
data['Transcripts'][0] = transcript

# Save the modified JSON
with open('kenny_transcripts_modified.json', 'w') as file:
    json.dump(data, file, indent=4)
```
