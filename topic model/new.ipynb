```python
import json
import gensim
from gensim import corpora
import pyLDAvis
import pyLDAvis.gensim
from nltk.corpus import stopwords
```

```python
# Load JSON data from file
with open('kanan_bag_of_words.json', 'r') as file:
    transcripts_data = json.load(file)
```

```python

# Extract transcripts
transcripts = transcripts_data

# Additional code for removing stopwords
stop_words = stopwords.words('english')
stop_words.extend(['let', 'like', 'one', 'yeah', 'two', 'okay', 'would', 'someone','say', 'go', 'guy', 'get', 'think', 'make', 'thing', 'know', 'come', 'look', 'see', 'tell', 'take', 'put', 'bro'])

def sent_to_words(sentences):
    for sentence in sentences:
        # deacc=True removes punctuations
        yield(gensim.utils.simple_preprocess(str(sentence), deacc=True))

def remove_stopwords(texts):
    return [[word for word in gensim.utils.simple_preprocess(str(doc)) 
             if word not in stop_words] for doc in texts]

# Tokenize the transcripts and create bigrams
tokens = list(sent_to_words(transcripts))
bigram = gensim.models.Phrases(tokens, min_count=5, threshold=100)
bigram_mod = gensim.models.phrases.Phraser(bigram)
tokens_bigram = [bigram_mod[doc] for doc in tokens]

# Remove stopwords
tokens_no_stop = remove_stopwords(tokens_bigram)
```

```python

# Create a dictionary and corpus
dictionary = corpora.Dictionary(tokens_no_stop)
corpus = [dictionary.doc2bow(doc) for doc in tokens_no_stop]

# Build the LDA model
lda_model = gensim.models.LdaModel(corpus=corpus, id2word=dictionary, num_topics=2, passes=10)

# Enable notebook display
pyLDAvis.enable_notebook()

# Create the visualization
vis = pyLDAvis.gensim.prepare(lda_model, corpus, dictionary, mds="mmds", R=10)

# Display the visualization
vis
```
